```python
import matplotlib.pyplot as plt

# Create a new figure and set its size (width, height)
plt.figure(figsize=(6,3))


# Store X-axis values (categories)
days = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]

# Store Y-axis values (number of steps for each day)
steps_walked = [300, 150, 500, 220, 125, 322, 654]


# Add label to X-axis
# labelpad = distance between label and axis
plt.xlabel("Days", labelpad=50)


# Add label to Y-axis
# labelpad = distance between label and axis
plt.ylabel("Steps Walked", labelpad=50)


# Add title to the graph
# pad = distance between title and graph
plt.title("My Step, Calories Count", pad=50)



# Plot the data as a line graph
plt.plot(
    days,                 # X-axis data
    steps_walked,         # Y-axis data

    marker="o",           # Put circle markers on data points
    markersize=10,        # Size of markers

    linestyle="--",       # Make line dashed
    linewidth=0.5,        # Thickness of line

    color="brown"         # Line color
)


# Display the graph
plt.show()
```

```python
import matplotlib.pyplot as plt


# Create a new figure and set graph size
plt.figure(figsize=(8,6))

# Store days for X-axis
days = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]

# Store steps data and calories data for first  and 2nd lines
steps_walked = [300, 150, 500, 220, 125, 322, 654]
calories_burned = [200, 172, 294, 185, 152, 216, 328]

# Add X-axis and Y-axis & Title's label
plt.xlabel("Days", labelpad=50)
plt.ylabel("(Steps Walked & Calories burned)", labelpad=50)
plt.title("My Step, Calories Count", pad=50)


# -------- First Line Graph --------
# Plot steps walked data
plt.plot(
    days,                  # X-axis values
    steps_walked,          # Y-axis values

    marker="o",            # Circle marker on points
    markersize=10,         # Size of circle

    linestyle="--",        # Dashed line style

    linewidth=0.5,         # Line thickness

    color="brown",         # Line color

    label="steps_walked"   # Name for legend
)



# -------- Second Line Graph --------
# Plot calories burned data
plt.plot(
    days,                    # X-axis values
    calories_burned,         # Y-axis values

    marker="^",              # Triangle marker

    markersize=14,           # Size of triangle

    linestyle="-",           # Solid line

    linewidth=5,             # Thickness of line

    color="pink",            # Line color

    label="calories_burned"  # Name for legend
)

# -------- Add Numbers on Points --------
# Loop through every day
for i in range(len(days)):

    # Write steps value above each step point
    plt.text(
        days[i],              # X position
        steps_walked[i] + 30,      # Y position
        steps_walked[i]       # Text to display
    )

    # Write calories value above each calorie point
    plt.text(
        days[i],
        calories_burned[i] -50,
        calories_burned[i]
    )



# Show labels of both lines
plt.legend()



# Add background grid
# True = show grid
# linestyle = grid line style
# alpha = transparency
plt.grid(True, linestyle="-", alpha=0.9)



# Display the final graph
plt.show()
```

```python
import matplotlib.pyplot as plt

days = ["Sun", "Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
steps_walked = [300, 150, 500, 220, 125, 322, 654]

# Step 1: build color list first
my_color = []

for i in range(len(days)):
    if steps_walked[i] > 250:
        my_color.append("red")
    else:
        my_color.append("green")

# Step 2: create figure
plt.figure(figsize=(5,3))

# Step 3: plot ONCE
plt.bar(days, steps_walked,
        edgecolor="black",
        color=my_color)

# Step 4: labels
plt.xlabel("Days")
plt.ylabel("Steps_Walked", labelpad=50)
plt.title("My Step Count", pad=10)

# Step 5: text on bars
for i in range(len(days)):
    plt.text(days[i], steps_walked[i] + 30,
             steps_walked[i],
             ha="center")

# Step 6: styling
plt.grid(True, linestyle="-")
plt.ylim(0, 850)

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(5,2))

days = ["Sun","Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
steps_walked = [300,150,500,220,125,322,654]

# Step 1: create colors first
my_color = ["green" if value > 250 else "red" for value in steps_walked]

# Step 2: plot bars
plt.bar(days, steps_walked,
        edgecolor="black",
        color=my_color)

# Step 3: add values on top of bars
for i in range(len(days)):
    plt.text(days[i], steps_walked[i] + 25,
             steps_walked[i],
             ha="center",
             fontsize=6)

# Labels
plt.xlabel("Days of the week")
plt.ylabel("Steps Walked")
plt.title("Bar Diagram 1st Practice")

plt.show()
```

```python
### PieChart
plt.figure(figsize=(6,3))
days = ["Sun","Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
steps_walked = [300,150,500,220,125,322,654]
my_wedge = {"linestyle": "--" , "edgecolor" : "darkblue"}
plt.title("Steps Walked", pad = 50)


plt.pie(steps_walked,
        labels = days,
        autopct = "%0.2f%%",
        textprops={"fontsize": 10},
        wedgeprops = my_wedge, shadow = True,
        pctdistance= 1.7)



plt.show()
```

```python

# Exploding the pie chart
### PieChart
plt.figure(figsize=(6,3))
days = ["Sun","Mon", "Tue", "Wed", "Thu", "Fri", "Sat"]
steps_walked = [300,150,500,220,125,322,654]

mean_steps = sum(steps_walked)/len(steps_walked)
print(mean_steps)

# my_explode = [0,0,0,0.1, 0,0,0.2]

my_explode = [0.2 if i > mean_steps else 0 for i in steps_walked]

#Calculate mean of steps_walked, and explode values less than mean by 0.2

my_wedge = {"linestyle": "--" , "edgecolor" : "darkblue"}
plt.pie(steps_walked,
        labels = days,
        autopct = "%0.2f%%",
        textprops={"fontsize": 10},
        wedgeprops = my_wedge, shadow = True,
        pctdistance= 1.7,
        explode = my_explode)


plt.show()
```
